```python
import numpy as np
import pandas as pd
```

```python
#series from list
```

```python
#string
country = ['India', 'Pakistan','USA','Nepal','Srilanka']
pd.Series(country)
```

```python
# integers
runs = [13,24,56,78,100]
runs_ser = pd.Series(runs)
```

```python
#custom index
marks = [67,57,89,100]
subjects = ['Maths', 'English', 'Science', 'Hindi']
pd.Series(marks,index=subjects)
```

```python
#setting a name
marks = pd.Series(marks,index=subjects,name = "Madhav results")
marks 
```

```python
# Series from dict
```

```python
marks = {
    'Maths':67,
    'English':57,
    'Science':89,
    'Hindi':100
}

marks= pd.Series(marks, name = 'madhav result')
marks
```

```python
# Series attrinutes
```

```python
# is_unique
#it tells if all the elements of a series is unique or not
marks.is_unique
#True
pd.Series([1,1,2,3,4,5]).is_unique
#False
```
